```python
# Act24 clase NaveEspacial Eliseo 1777
# Registro interactivo V2 - 4 atributos y 3 funciones

class NaveEspacial:
    def __init__(self, nombre, tripulacionMaxima, combustible, mision):
        self.nombre = nombre                # Atributo 1
        self.tripulacionMaxima = tripulacionMaxima  # Atributo 2
        self.combustible = combustible      # Atributo 3
        self.mision = mision                # Atributo 4

    def despegar(self):
        return f"🚀 La nave {self.nombre} ha iniciado el despegue para la misión: {self.mision}."

    def orbitar(self):
        return f"🌍 {self.nombre} está en órbita estable con {self.combustible}% de combustible."

    def transmitirDatos(self):
        return f"📡 [SEÑAL] Transmitiendo datos de la tripulación ({self.tripulacionMaxima} personas) a la Tierra..."

# --- PROGRAMA FUNCIONANDO INTERACTIVO ---
print("=== CENTRO DE CONTROL ESPACIAL - REGISTRO V2 ===")

# Registro de la primera nave
print("\nDATOS DE LA NAVE 1:")
nom1 = input("Nombre de la nave: ")
tri1 = input("Tripulación máxima: ")
com1 = input("Nivel de combustible (%): ")
mis1 = input("Objetivo de la misión: ")
nave1 = NaveEspacial(nom1, tri1, com1, mis1)

# Registro de la segunda nave
print("\nDATOS DE LA NAVE 2:")
nom2 = input("Nombre de la nave: ")
tri2 = input("Tripulación máxima: ")
com2 = input("Nivel de combustible (%): ")
mis2 = input("Objetivo de la misión: ")
nave2 = NaveEspacial(nom2, tri2, com2, mis2)

# --- RESULTADOS ---
for n in [nave1, nave2]:
    print("\n" + "="*50)
    print(f"REPORTE TÉCNICO: {n.nombre}")
    print(f"Misión: {n.mision} | Capacidad: {n.tripulacionMaxima} | Energía: {n.combustible}%")
    print("-" * 50)
    print(n.despegar())
    print(n.orbitar())
    print(n.transmitirDatos())

print("\n" + "="*50)
print("Programa realizado por Marco Morquecho NL=39")
```
